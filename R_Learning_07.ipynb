```r
# Reading SPSS file
library(haven)
data_spss <- read_sav("/content/dataset.sav")
write_sav(data_spss, "/content/dataset.sav")

# Converting character variable/column as factor
data_spss$Gender <- as.factor(data_spss$Gender)
# or
data_spss %>%
  mutate(across(where(is.character), as.factor)) %>%
  write_sav("/content/dataset.sav")
```

```r
# Reading STATA file
library(haven)
data_stata <- read_dta("/content/dataset.sav")
write_dta(data_stata, "/content/dataset.sav")

```

```r
# Reading JSON file
library(jsonlite)
data_json <- fromJSON("/content/dataset.sav")
data_json %>%
  mutate(Year = as.integer(Year),
  values = as.numeric(values)) %>%
  write_json("/content/dataset.sav", pretty = TRUE)
```

```r
data(population, package = "tidyr")
population <- data.frame(population)
data(who, package = "tidyr")
who <- data.frame(who)

install.packages("RSQLite")
library(DBI)      # contains functions to interact with database
library(RSQLite)   #SQLite database driver
```

```r
# Create an in-memory SQLite database connection within a database server
con <- dbConnect(drv = SQLite(), ":memory:")

# List all tables in the database
dbListTables(con)

# Write the population and who dataframes to the database
dbWriteTable(conn = con, name = 'pop', value = population)
dbWriteTable(conn = con, name = 'who', value = who, overwrite = TRUE)


# Append a subset of mtcars dataframe to a new table
dbWriteTable(con, "mtcars", mtcars[1:5,], append = TRUE)

# Remove the original dataframes from the environment
rm(populaiton, who)

# Retrive data from the who table using SQL
dbGetQuery(
  conn = con,
  statement = "
  SELECT iso2 AS country, year, new_sp_m014, new_sp_f014
  FROM who
  WHERE country IN ('Bangladesh', 'India') AND year >=2000 AND year<=2006
  ")

  # Retrive data from table who using dplyr
  tbl(con, "who") %>%
    filter(country %in% c('Bangladesh','India') & year >= 2000 & year <=2006) %>%
    select(iso2, year, new_sp_m014, new_sp_f014) %>%
    collect()   # Retrives the data into a local tribble

  # Disconnect from the server
  dbDisconnect(conn)
```
